```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_curve, roc_auc_score
```

```python
X, y = make_classification(
    n_samples=1200,
    n_features=20,
    n_informative=15,
    n_classes=2,
    random_state=42
)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
model_lr = LogisticRegression(max_iter=1000, random_state=42)
model_rf = RandomForestClassifier(n_estimators=100, random_state=42)
```

```python
model_lr.fit(X_train, y_train)
model_rf.fit(X_train, y_train)
```

```python
# Note: [:, 1] extracts probabilities for the positive class
lr_probs = model_lr.predict_proba(X_test)[:, 1]
rf_probs = model_rf.predict_proba(X_test)[:, 1]
```

```python
lr_auc = roc_auc_score(y_test, lr_probs)
rf_auc = roc_auc_score(y_test, rf_probs)
```

```python
print(f"Logistic Regression AUC: {lr_auc:.4f}")
print(f"Random Forest Classifier AUC: {rf_auc:.4f}")
```

```python
# Calculate ROC curve coordinates (FPR, TPR) for both models
lr_fpr, lr_tpr, _ = roc_curve(y_test, lr_probs)
rf_fpr, rf_tpr, _ = roc_curve(y_test, rf_probs)
```

```python
plt.figure(figsize=(8, 6))

# Plot the model lines
plt.plot(lr_fpr, lr_tpr, linestyle='--', color='orange', label=f'Logistic Regression (AUC = {lr_auc:.3f})')
plt.plot(rf_fpr, rf_tpr, linestyle='-', color='green', label=f'Random Forest (AUC = {rf_auc:.3f})')

# Plot the baseline (Random Classifier) diagonal line
plt.plot([0, 1], [0, 1], linestyle=':', color='blue', label='Random Guess (AUC = 0.500)')

# Graph Customization
plt.title('ROC Curve Model Comparison')
plt.xlabel('False Positive Rate (FPR)')
plt.ylabel('True Positive Rate (TPR)')
plt.legend(loc='lower right')
plt.grid(True, linestyle=':', alpha=0.6)

# Display the plot
plt.show()
```

# Regression 

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

## 1. Simple Linear Regression

```python
X = np.array([[1], [2], [3], [4], [5]]) 
y = np.array([45, 50, 60, 75, 90])

model = LinearRegression()

model.fit(X, y)

# Predict score for 6 hours of study
prediction = model.predict([[6]])
print(f"Predicted Score: {prediction[0]}")
```

## 2. Multiple Linear Regression

```python
X = [[1500, 3], [2000, 4], [1200, 2], [2500, 5]]
y = [300000, 400000, 250000, 500000]

model = LinearRegression()

model.fit(X, y)

# Predict price for 1800 sq ft and 3 rooms
new_house = [[1800, 3]]
print(f"Predicted Price: ${model.predict(new_house)[0]:,.2f}")
```

## 3. Polynomial Regression

```python
from sklearn.preprocessing import PolynomialFeatures

X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([1, 4, 9, 16, 25]) # Quadratic relationship

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

model = LinearRegression()

model.fit(X_poly, y)

print(f"Model Coefficients: {model.coef_}")
```

## Ridge Regression

```python
from sklearn.linear_model import Ridge

# Sample Data
X = np.array([[1, 1], [1, 2], [2, 2], [2, 3]])
# y = 1 * x_0 + 2 * x_1 + 3
y = np.dot(X, np.array([1, 2])) + 3


ridge_model = Ridge(alpha=0.1)
ridge_model.fit(X, y)

print(f"Coefficients: {ridge_model.coef_}")
print(f"Intercept: {ridge_model.intercept_}")

# Predicting a new value
print(f"Prediction: {ridge_model.predict([[3, 5]])}")
```

## Gradient Descent via SGDRegressor

```python
from sklearn.linear_model import SGDRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

# Sample Data (e.g., Square Footage and Age of House)
X = np.array([[1200, 10], [1500, 5], [2000, 15], [800, 2], [2500, 20]])
y = np.array([250000, 310000, 380000, 190000, 450000])

model = make_pipeline(StandardScaler(), SGDRegressor(max_iter=1000, alpha=0.01, tol=1e-3, loss='squared_error', penalty='l2'))

model.fit(X, y)

# Predict
print(f"Predicted Value: {model.predict([[1800, 8]])[0]:.2f}")
print(f"Predicted Value: {model.predict([[1200, 10]])[0]:.2f}")
```
